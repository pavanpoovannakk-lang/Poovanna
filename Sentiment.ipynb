```python
from transformers import pipeline

# Load a pre-trained sentiment analysis model
model_name = "nlptown/bert-base-multilingual-uncased-sentiment"

# Initialize a sentiment analysis pipeline
sentiment_pipeline = pipeline("sentiment-analysis", model=model_name)

# Define a mapping from 5-star labels to 3-category sentiment
sentiment_mapping = {
    '1 star': 'negative',
    '2 stars': 'negative',
    '3 stars': 'neutral',
    '4 stars': 'positive',
    '5 stars': 'positive'
}

for i in range(3):
    input_sentence = input(f"Please enter sentence {i+1} for sentiment analysis: ")

    # The pipeline handles tokenization, model inference, and result decoding
    result = sentiment_pipeline(input_sentence)

    sentiment_label_5_stars = result[0]['label']
    score = result[0]['score']

    # Map the 5-star label to a 3-category sentiment
    classified_sentiment = sentiment_mapping.get(sentiment_label_5_stars, 'unknown')

    print(f"The sentiment of sentence {i+1} is: {classified_sentiment} (original: {sentiment_label_5_stars}, score: {score:.2f})")
```
